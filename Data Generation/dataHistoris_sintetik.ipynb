```python
import pandas as pd
import random
from datetime import datetime, timedelta

df_game = pd.read_csv('dimGame_cleaned.csv')
df_user = pd.read_csv('name.csv')
```

```python
#generate tanggal random
start_date = datetime(2022, 1, 1)
end_date = datetime(2026, 1, 1)
def random_date(start, end):
    delta = end - start
    random_days = random.randrange(delta.days)
    random_seconds = random.randrange(24*60*60)
    return start + timedelta(days=random_days, seconds=random_seconds)

genre_columns = df_game.columns[6:].tolist() 
data_transaksi = []

for index, user in df_user.iterrows():
    user_id = user['cust_id'] # Ganti dengan nama kolom ID Anda
    #tentuin genre favorit user
    num_genres = random.randint(5, 8)
    user_fav_genres = random.sample(genre_columns, k=num_genres)
    #tentuin platform
    persentase = [0.30, 0.30, 0.20, 0.10, 0.10]
    daftar_platform = ['PS5','PS4','PS3','PSP','PS Vita']
    user_platform = random.choices(daftar_platform, weights=persentase, k=1)[0]
    df_platform_games = df_game[df_game['platform'] == user_platform]

    # Jika dijumlahkan hasilnya > 0, berarti game itu memiliki minimal 1 genre favorit user
    jumlah_kesamaan = df_platform_games[user_fav_genres].sum(axis=1)
    is_fav_game = (jumlah_kesamaan >= 1) & (jumlah_kesamaan <= 2)
    fav_games = df_platform_games[is_fav_game]
    other_games = df_platform_games[~is_fav_game]
    num_games_owned = random.randint(30, 40)
    
    for _ in range(num_games_owned):
        # 85% user memilih game sesuai genre favorit
        # 15% user memilih diluar favorit
        if random.random() < 0.85 and not fav_games.empty:
            chosen_game = fav_games.sample(1).iloc[0]
            #supaya 1 user tidak membeli game yang sama
            fav_games = fav_games.drop(chosen_game.name)
        elif not other_games.empty:
            chosen_game = other_games.sample(1).iloc[0]
            other_games = other_games.drop(chosen_game.name)
        elif not fav_games.empty: 
            chosen_game = fav_games.sample(1).iloc[0]
            fav_games = fav_games.drop(chosen_game.name)
        else:
            break
        data_transaksi.append({
            'user_id' : f"{int(user['cust_id']):04d}",
            'game_name': chosen_game['game_name'],
            'platform': chosen_game['platform'],
            'user_profile_genres': " | ".join(user_fav_genres),
            'playtime_hours': round(random.uniform(5.0, 200.0), 1), 
            'rating': random.randint(4, 5) if random.random() < 0.75 else random.randint(1, 3),
            'date_time': random_date(start_date, end_date)
        })
    if (index + 1) % 50 == 0:
        print(f"Sedang memproses...User ke-{index + 1} | Total transaksi: {len(data_transaksi)}")

# Jadikan DataFrame baru
df_history = pd.DataFrame(data_transaksi)
df_history.head(50)
```

```python
df_history.to_csv('synthetic_trans.csv', index=False)
print("Data transaksi berhasil disimpan ke synthetic_transaction_history.csv!")
```
