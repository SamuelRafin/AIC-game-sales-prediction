```python
import numpy as np
import pandas as pd
import random
df = pd.read_csv("nameExtractor.csv", sep=',', encoding='utf-8')
df.head(10)
```

```python
#name
first_name = df["First Name"].dropna().tolist()
last_name = df["Last Name"].dropna().tolist()

pengali_first = (5000 // len(first_name)) + 1
pengali_last = (5000 // len(last_name)) + 1

first_names_extended = (first_name * pengali_first)[:5000]
last_names_extended = (last_name * pengali_last)[:5000]

random.shuffle(first_names_extended)
random.shuffle(last_names_extended)

hasil = []
for i in range(5000):
    fn = first_names_extended[i]
    ln = last_names_extended[i]
    hasil.append({
        "cust_id": f"{i+1:04d}",
        "First Name": fn,
        "Last Name": ln,
        "Full Name": f"{fn} {ln}"
    })

#contact
nomor_telepon = []
for _ in range(5000):
    acak = "".join([str(random.randint(0, 9)) for _ in range(10)])
    nomor = "08" + acak
    nomor_telepon.append(nomor)
df_telepon = pd.DataFrame(nomor_telepon, columns=["Nomor Telepon"])

df_final = pd.DataFrame(hasil)
df_final["Nomor Telepon"] = df_telepon["Nomor Telepon"]
df_final.head(10)
```

```python
df_final.to_csv("name.csv", index=False, encoding='utf-8')
```
